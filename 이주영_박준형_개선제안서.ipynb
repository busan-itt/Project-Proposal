# 부산항 효율 제고 제안서


## 주제 : 선석 통합운영을 통한 부산 신항 환적 체류시간 저감

---

## 요약

부산 신항은 터미널별로 선석을 독립적으로 운영하고 있어 특정 터미널에 물량이 집중되더라도 다른 터미널의 여유 자원을 유연하게 활용하는 데 한계가 있다. 자부두·타부두 환적 데이터 분석에서는 야드 체류가 장시간 지속되는 현상과 환적 방향별 체류시간의 차이가 확인되었으나, 선박 운영시간과 야드 체류시간의 상관분석만으로 체류 원인을 특정하기는 어려웠다.

본 제안서는 이러한 분석 결과를 바탕으로 부산 신항 7개 터미널의 선석을 공동으로 활용하는 개선 방안을 제시한다. 기존 운영, 인접 터미널 간 선석 공유, 신항 전체 통합운영을 시뮬레이션으로 비교하여 야드 체류시간, 선박 대기시간 및 터미널별 부하율의 개선 가능성을 검토한다.

## 1. 현황 및 문제점

### 1) 현재 방식 및 절차

- 자부두 환적은 동일 터미널 내에서 컨테이너를 양하한 뒤 야드에 장치하고, 연결선의 적하 일정에 맞춰 온샤시 및 반출하는 방식으로 이루어진다.

- 타부두 환적은 출발 터미널의 반출 이후 다른 터미널로 이동하여 다시 반입·장치·온샤시·반출하는 공정이 추가된다. 이에 따라 출발 터미널과 도착 터미널의 야드 체류 및 선박 연결 일정을 함께 고려할 필요가 있다.

- 부산 신항의 PNC, PNIT, HPNT, HJNC, BNCT, BCT, DGT는 터미널별로 선석을 운영하고 있다. 개별 터미널 중심의 배정 방식에서는 특정 터미널에 입항이 집중되더라도 다른 터미널의 빈 선석을 즉시 활용하기 어려울 수 있다.


### 2) 발견된 문제점

#### (1) 환적 컨테이너의 장기 야드 체류

- HJNC 자부두 환적 컨테이너 1,490건을 분석한 결과, 평균 야드 체류시간은 195.9시간, 중앙값은 168.9시간으로 나타났다. 야드 체류시간은 야드 장치부터 온샤시까지의 시간으로 계산하였다.

- 120시간 이상 체류한 컨테이너는 1,093건으로 전체의 73.4%를 차지하였다. 일부 극단 사례뿐 아니라 다수의 컨테이너에서 장시간 체류가 나타나 체류 구조를 검토할 필요가 있다.

- 다만 야드 체류에는 정상적인 연결선 대기가 포함될 수 있으므로, 장시간 체류 전체를 운영 지연으로 단정하지 않는다.

#### (2) 환적 방향별 야드 체류시간의 차이

- HJNC–DGT 타부두 환적 컨테이너 38,196건의 분석에서는 DGT→HJNC의 평균 총 야드 체류시간이 282.81시간, HJNC→DGT는 176.86시간으로 나타났다. 총 야드 체류시간은 두 터미널의 장치부터 온샤시까지의 시간을 합산한 값이다.

- 두 방향 모두 출발 터미널에서 평균적으로 더 오래 체류하였다. DGT→HJNC에서는 일부 초장기 체류 기록이 평균에 큰 영향을 주었으며, 해당 기록을 제외한 비교에서도 방향 간 차이가 남았다.

- 이러한 결과는 터미널 간 이동 공정뿐 아니라 각 터미널의 야드 체류 및 선박 연결 조건을 함께 검토할 필요가 있음을 보여준다.

#### (3) 개별 선박 시간 지표만으로 설명하기 어려운 체류 구조

- HJNC 자부두 환적 분석에서는 현재 선박의 시간 지표 5개와 야드 체류시간 간 상관계수의 절댓값이 모두 0.12 미만이었다.

- 타부두 환적 분석에서도 컨테이너 기준과 항차 기준 모두 뚜렷한 선형 관계를 확인하기 어려웠다. DGT 자부두 환적 분석에서는 본선작업완료지연시간과 항차별 TEU 가중평균 야드 장치 기간 사이에 유의수준 0.05에서 유의한 선형 상관을 확인하지 못하였다.

- 따라서 개별 선박의 작업시간이나 대기시간을 줄이면 야드 체류시간도 같은 비율로 줄어든다고 단정하기 어렵다. 선석 배정과 선박 간 연결 조건을 변경하여 변화 가능성을 직접 비교할 필요가 있다.

### 3) 개선 필요성

- 야드 체류시간의 분포와 환적 방향별 차이를 확인한 현황 분석에서 나아가, 운영방식 변경이 실제로 어떤 지표를 개선할 수 있는지 검토할 필요가 있다.

- 특정 터미널의 혼잡을 개별 터미널 내부에서만 해결하는 방식에서 벗어나, 신항 전체의 선석과 처리능력을 공동으로 활용하는 방안을 검토한다.

- 선석 통합운영이 야드 체류시간과 선박 대기시간을 줄이고, 과부하 터미널의 부담을 완화할 수 있는지 시뮬레이션으로 확인한다.

## 2. 개선 방안

### 1) 구체적 개선 내용

#### (1) 터미널별 운영자료를 연계한 선석 배정 기준 마련

- 7개 터미널의 선석배정현황과 선박 제원을 연계하여 선박별 입항·접안·출항 및 선석 점유 정보를 정리한다.

- 선박 길이, 안벽 길이, 시간대별 점유 상태 등 확보 가능한 조건을 기준으로 배정 가능한 선석을 구분한다. 수심, 접안 능력 및 하역 설비 등 추가 조건은 자료 확보 범위에 따라 반영한다.

- 연간 하역능력과 처리 물량은 터미널별 부하율 평가에 활용한다. 연간 하역능력을 개별 선박의 시간당 작업속도와 동일하게 취급하지 않는다.

#### (2) 선석 공유 및 통합운영 시나리오 구성
| 구분 | 운영방식 | 비교 목적 |
| :--- | :--- | :--- |
| 기존 운영 | 선박을 기존 터미널의 선석에 배정 | 현재 운영을 비교 기준으로 설정 |
| 인접 터미널 공유 | 기존 터미널과 인접 터미널의 선석을 공동 활용 | 제한적인 선석 공유의 개선 가능성 확인 |
| 신항 전체 통합 | 7개 터미널의 선석을 공동 활용 | 통합 범위 확대에 따른 개선 가능성 확인 |

- 입항이 집중된 터미널의 선박을 배정 가능한 다른 터미널의 여유 선석으로 분산하는 방안을 검토한다.

- 실제 터미널 배치와 이동 조건을 확인하여 인접 공유 범위를 설정하고, 시나리오 간에는 동일한 수요와 비교 기준을 적용한다.

#### (3) 환적 연결 조건을 고려한 운영방안 검토

- 컨테이너의 양하모선과 적하 연결선을 연계하여 선석 배정 변경이 환적 연결 간격 및 야드 체류시간에 미치는 영향을 비교한다.

- 모선과 연결선이 함께 영향을 받을 수 있으므로, 한 선박의 접안이 빨라지는 효과만으로 전체 체류 감소를 판단하지 않는다.

- 필요한 경우 연결선 일정 조정, 배정 우선순위 및 최소 환적 소요시간을 추가 조건으로 설정하여 통합운영 내에서 적용 가능한 방안을 검토한다. 조정 범위와 실행 가능성은 별도로 확인한다.


#### (4) 시뮬레이션 검증 및 보완

- 기존 운영을 재현하는 모델을 구성하고 관측 자료와 비교하여 재현 수준을 확인한다.

- 운영 시나리오별로 야드 체류시간, 선박 대기시간 및 터미널별 부하율을 비교한다.

- 반복 실험과 민감도 분석을 통해 선박 제원 결측, 배정 조건 및 시간 가정이 결과에 미치는 영향을 확인한다.

- 자료가 충분한 경우 기간을 구분한 검증을 수행하고, 결측 제원 및 선석 점유시간 보완이 필요한 경우 예측모델 활용을 검토한다.

### 2) 적용 범위

- 선석 운영 검토 대상: 부산 신항 7개 터미널(PNC, PNIT, HPNT, HJNC, BNCT, BCT, DGT).

- 환적 체류 검토 대상: 확보된 HJNC·DGT 자부두 환적 및 HJNC↔DGT 타부두 환적 자료.

- 분석 기준: 2025년 자료를 활용하되, 자료별 전처리 조건과 시간 정의를 확인하여 비교 가능한 범위를 설정한다.

- 적용 방식: 실제 운영 변경에 앞서 과거 자료를 활용한 가상 운영 시나리오로 검증한다. 환적 표본에서 산출한 결과를 신항 전체 화물의 효과로 확대하지 않는다.

## 3. 기대 효과

### 1) 정량적 효과

본 제안서는 다음 세 가지 효과를 중심으로 개선 가능성을 평가한다. 구체적인 감소량과 감소율은 기존 운영과 개선 시나리오를 동일 조건에서 비교한 후 산출한다.

#### (1) 야드 체류시간 감소

- 환적 연결 관계를 고려한 선석 배정 및 운영 일정 관리를 통해 컨테이너가 야드에서 다음 선박을 기다리는 시간을 줄일 수 있을 것으로 기대한다.


- 야드 체류시간은 장치부터 온샤시까지의 시간을 기준으로 하며, 타부두 환적은 두 터미널의 야드 체류시간을 합산한다. 자료별 시간 정의가 다른 경우 비교 전에 기준을 정리한다.

- 평가 지표: 평균 야드 체류시간, 기존 대비 감소시간 및 감소율.

- 감소율 산정: (기존 평균 야드 체류시간 − 개선 후 평균 야드 체류시간) / 기존 평균 야드 체류시간 × 100.

####(2) 선박 대기시간 감소

- 특정 터미널에 선박이 집중되는 경우 다른 터미널의 여유 선석을 활용함으로써 접안까지 발생하는 대기시간을 줄일 수 있을 것으로 기대한다.

- 평가 지표: 평균 접안 대기시간, 기존 대비 감소시간 및 감소율.

- 선박 대기시간은 배정 가능 상태가 된 시점부터 접안까지의 시간으로 정의한다. 입항 후 작업 시작까지의 대기 및 반입 시작·마감 기준 경과시간과 구분한다. 실제 도착 자료가 부족한 경우 대체 기준을 명시한다.

#### (3) 터미널별 부하율 감소

- 물량이 집중된 터미널의 선박과 처리 물량을 여유 터미널로 분산하여 과부하 터미널의 부하율을 낮추고 터미널 간 운영 불균형을 완화할 수 있을 것으로 기대한다.

- 평가 지표: 터미널별 부하율, 최대 부하율 및 터미널 간 부하율 편차.

- 부하율 산정: 터미널별 연간 처리 물량 / 해당 터미널의 연간 하역능력 × 100.

- 물량을 분담하는 터미널에서는 부하율이 높아질 수 있으므로, 모든 터미널의 부하율이 동시에 감소하는 것을 목표로 설정하지 않는다. 과부하 완화와 편차 축소를 중심으로 평가한다.

### 2) 정성적 효과

- 야드 체류시간 관리: 장기 체류 현황과 환적 연결 조건을 함께 파악하여 체류 관리의 일관성 제고.

- 선박 대기 관리: 터미널별 여유 선석 정보를 공동으로 활용하여 혼잡 발생 시 대응 유연성 향상.

- 터미널 부하 관리: 개별 터미널 중심의 배정에서 항만 전체의 자원 활용을 고려하는 운영 기반 마련.

### 3) 효과 검증 및 활용

- 세 지표의 개선 여부와 크기는 시뮬레이션을 통해 확인한다. 지표 간 효과가 다르게 나타나는 경우 적용 조건과 상충관계를 함께 비교하여 개선안을 보완한다.
